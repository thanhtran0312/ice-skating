```python
objPoints_total = np.array(objPoints_total, dtype=np.float32)  
imgPoints_total = np.array(imgPoints_total, dtype=np.float32)  
```

```python
# 3. Camera intrinsics 
K = np.array([[3.54548296e+03, 0.00000000e+00, 9.60287866e+02],
                [0.00000000e+00, 3.54924565e+03, 5.57450277e+02],
                [0.00000000e+00, 0.00000000e+00, 1.00000000e+00]])
distortion = np.array([5.22080106e-02,5.61495738e+00,1.01498704e-02,2.81631980e-03,-2.88237239e+01])
```

```python
# 4. Solve for extrinsics
ret, rvec, tvec = cv2.solvePnP(
    objPoints_total,
    imgPoints_total,
    K,
    distortion,
    flags=cv2.SOLVEPNP_ITERATIVE
)
cv2.solvePnPRefineLM(
    objPoints_total,
    imgPoints_total,
    K,
    distortion,
    rvec,
    tvec
)
print("Rotation vector (rvec):", rvec)
print("Translation vector (tvec):", tvec)

R, _ = cv2.Rodrigues(rvec)
print("Rotation matrix (R):\n", R)
```

```python
# reprojection error
# reprojection error
proj_points, _ = cv2.projectPoints(
    objPoints_total, rvec, tvec, K, distortion
)

proj_points = proj_points.reshape(-1, 2)

error = np.linalg.norm(proj_points - imgPoints_total, axis=1)
print("Reprojection error per point (px):", error)
print("Mean reprojection error (px):", error.mean())

```

```python
ret, rvec, tvec, inliers = cv2.solvePnPRansac(
    objPoints_total,
    imgPoints_total,
    K,
    distortion,
    None,
    None,
    False,
    8,
    20.0,
    0.99,
    
    cv2.SOLVEPNP_ITERATIVE
)

objPoints_inliers = objPoints_total[inliers[:,0]]
imgPoints_inliers = imgPoints_total[inliers[:,0]]
cv2.solvePnPRefineLM(objPoints_inliers, imgPoints_inliers, K, distortion, rvec, tvec)
```

```python
# reprojection error
# reprojection error
proj_points, _ = cv2.projectPoints(
    objPoints_inliers, rvec, tvec, K, distortion
)

proj_points = proj_points.reshape(-1, 2)

error = np.linalg.norm(proj_points - imgPoints_inliers, axis=1)
print("Reprojection error per point (px):", error)
print("Mean reprojection error (px):", error.mean())

```
